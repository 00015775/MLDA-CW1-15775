# 1. Data Loading & Feature Descriptions

````python
# relative path
# out of 21 features in the original dataset, only 7 of them were selected

crime_df = pd.read_csv(crime_path, 
                       usecols=["Latitude", "Longitude", "CrimeDateTime", "Inside_Outside", "PremiseType", "Description"],
                       sep=",")

# adjusts data type and extract datatime details, this step also effectively extracts and adds news features
crime_df["CrimeDateTime"] = pd.to_datetime(crime_df["CrimeDateTime"])
crime_df["Hour"] = crime_df["CrimeDateTime"].dt.hour
crime_df["Month"] = crime_df["CrimeDateTime"].dt.month_name()
crime_df["DayOfWeek"] = crime_df["CrimeDateTime"].dt.day_name() # 0=Monday, 1=Tuesday, ... 6=Saturday

# now for self-preference adjust the order of columns, where they are ordered based on the relevance and similarity
# target column is at the right while all input features are on the left
desired_order = ["Latitude", "Longitude", "CrimeDateTime", "Month", "DayOfWeek", "Hour", "PremiseType", "Inside_Outside", "Description"]
crime_df = crime_df[desired_order]
````

## dataset splitting training/testing sets

````python
# Target variable is Description of the type of Crime occuring based on the location and time
X = crime_df_clean_encoded.drop('Description', axis=1)
y = crime_df_clean_encoded['Description']

# combining the dataset to apply changes to both of them
imbalance_df = X.copy()
imbalance_df['Description'] = y
class_counts = imbalance_df['Description'].value_counts()

# Only the entries with at least 10 samples from each column will be kept
min_samples = 10

# rows with entires that have at least 10 samples are kept, the rest are disregarded
valid_classes = class_counts[class_counts >= min_samples].index
df_filtered = imbalance_df[imbalance_df['Description'].isin(valid_classes)]

print("Classes kept:", df_filtered['Description'].nunique())

num_classes = df_filtered['Description'].nunique() # select only distinct classes
samples_per_class = 200000 // num_classes 

print(f"Samples per class to get 200k rows: {samples_per_class}")
````

````python
balanced_df_list = []

for cls in valid_classes:
    cls_df = df_filtered[df_filtered['Description'] == cls]
    
    # If class has fewer rows than samples_per_class, do upsampling
    if len(cls_df) < samples_per_class:
        cls_resampled = resample(cls_df,
                                 replace=True,       
                                 n_samples=samples_per_class,
                                 random_state=42)
    else:
        cls_resampled = cls_df.sample(n=samples_per_class, random_state=42)
    
    balanced_df_list.append(cls_resampled)

# Combine all classes
df_balanced = pd.concat(balanced_df_list).sample(frac=1, random_state=42)
````

````python
X_balanced = df_balanced.drop('Description', axis=1)
y_balanced = df_balanced['Description']

print("Balanced dataset shape:", X_balanced.shape)
print("Class distribution:\n", y_balanced.value_counts())
````

````python
# 80% to training and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X_balanced, y_balanced, test_size=0.2, random_state=42
)
````

````python
# shapes
print(f"\nX_train shape: {X_train.shape}, X_test shape: {X_test.shape}\n")
print(f"y_train shape: {y_train.shape}, y_test shape: {y_test.shape}\n")
````

# 4. Models Training

## $1^{st}$ RandomForest (RF) algorithm

````python
# Hyperparameter tuning for n_estimators

rf_param_options = {
    'n_estimators': [50, 100, 200, 300],
    'min_samples_leaf': [1, 2, 4],
    'class_weight': ['balanced']
}

rf_md = RandomForestClassifier(random_state=42)

# Fitting RF model
rf_cv = GridSearchCV(rf_md, rf_param_options, cv=5, scoring="f1_macro") # f1_macro prevents large classes from dominating
rf_cv.fit(X_train, y_train)

print(f"The best n_estimators hyperparameter for Random Forest: {rf_cv.best_params_}")
````

````python
# Train RF classifier with best params
rf_model = RandomForestClassifier(
    #n_estimators=rf_cv.best_params_['n_estimators'], 
    n_estimators=500, # because the grid search was taking a lot of time, I decided to manually experiment with higher values, 500 provided better accuracy
    random_state=42,
    class_weight='balanced'
)

# Fitting the model
rf_model.fit(X_train, y_train)

# Predictions
y_pred_rf = rf_model.predict(X_test)

print(y_pred_rf)
print(classification_report(y_test, y_pred_rf, zero_division=0))
````

````python
models = {
    "Random Forest": rf_model,
    "HistGradientBoosting": hgb_model,
    "CatBoost": cb_model
}

# Evaluate each model
for name, model in models.items():
    print(f"\n===== {name} =====")
    
    # Predict on test set
    y_pred = model.predict(X_test)
    
    # Compute metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, average='macro', zero_division=0)
    rec = recall_score(y_test, y_pred, average='macro', zero_division=0)
    f1 = f1_score(y_test, y_pred, average='macro', zero_division=0)
    
    # Print results
    print(f"Accuracy: {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall: {rec:.4f}")
    print(f"F1-score: {f1:.4f}")
````
